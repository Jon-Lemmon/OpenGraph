#### Geração de Chave SSH
É sempre preciso rodar ao início de uma sessão do Colab e colar a chave pública no GitHub.

```python
# Certifique-se de estar na pasta correta antes de executar as células abaixo.
# Gere a chave numa pasta *fora* do repositório.
!ls
```

```python
# Definindo caminhos do projeto
PROJECT_DIR = Path.cwd()
SSH_DIR = PROJECT_DIR / '.ssh'

PROJECT_DIR.mkdir(parents=True, exist_ok=True)
SSH_DIR.mkdir(parents=True, exist_ok=True)

REPO_DIR = PROJECT_DIR / 'OpenGraph'
KEY_FILE = SSH_DIR / 'github_ed25519'
CONFIG_FILE = SSH_DIR / 'config'
KNOWN_HOSTS = SSH_DIR / 'known_hosts'
```

```python
if not KEY_FILE.exists():
    subprocess.run([
        'ssh-keygen',
        '-t', 'ed25519',
        '-C', 'colab-opengraph',
        '-f', str(KEY_FILE),
        '-N', ''
    ], check=True)

# Pasta onde a chave será salva: './.ssh'
print(KEY_FILE.with_suffix('.pub').read_text())
```

```python
!ls .ssh
```

```python
KNOWN_HOSTS.touch(exist_ok=True)

config = f"""
Host github.com
    HostName github.com
    User git
    IdentityFile {KEY_FILE.resolve()}
    IdentitiesOnly yes
    UserKnownHostsFile {KNOWN_HOSTS.resolve()}
    StrictHostKeyChecking accept-new
"""

CONFIG_FILE.write_text(config)
```

```python
os.environ['GIT_SSH_COMMAND'] = f'ssh -F "{CONFIG_FILE.resolve()}"'
```

```python
!ssh -F "{CONFIG_FILE.resolve()}" -T git@github.com
```
